**LIMPIEZA DEL DATASET**

```python
import pandas as pd
ruta= '/content/supply_chain_logistics_delay_dataset.csv'
df= pd.read_csv(ruta)

df.info(verbose=True)
```

```python
df.head()
```

```python
columnas_relevantes = [
    # Almacén y operaciones
    'warehouse_type', 'warehouse_capacity', 'warehouse_utilization',
    'staffing_level', 'order_backlog', 'backorder_quantity',

    # Inventario
    'inventory_level', 'inventory_turnover', 'stockout_risk',

    # Tiempos operativos
    'processing_time_hours', 'picking_time_hours', 'packing_time_hours',
    'loading_time_hours', 'handling_time_hours',

    # Costos
    'warehouse_cost', 'handling_cost', 'transportation_cost',
    'customs_cost', 'insurance_cost', 'total_logistics_cost',

    # Resultado / entregas
    'estimated_delivery_days', 'actual_delivery_days',
    'delay_days', 'delivery_delayed', 'delivery_performance_score',

    # Contexto del envío
    'product_category', 'order_quantity', 'order_value',
    'package_weight_kg', 'package_volume_m3', 'distance_km'
]

df_almacenes = df[columnas_relevantes].copy()
print(f"Dataset original: {df.shape[1]} columnas")
print(f"Dataset nuevo: {df_almacenes.shape[1]} columnas")
df_almacenes.head()
```
